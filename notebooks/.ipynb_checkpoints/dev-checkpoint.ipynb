```python
import os

def load_dataset(base_path):
    data = []
    
    for label in os.listdir(base_path):
        label_path = os.path.join(base_path, label)
        
        if not os.path.isdir(label_path):
            continue
        
        for file in os.listdir(label_path):
            img_path = os.path.join(label_path, file)
            
            img = cv2.imread(img_path)
            
            if img is not None:
                data.append({
                    "image": img,
                    "label": label
                })
    
    return data
```

```python
from pathlib import Path
from copy import deepcopy
import numpy as np
 
# ── Reproducibility ──────────────────────────────────────────
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
 
# ── Dataset paths (adjust if needed) ─────────────────────────
BASE_DIR = Path(r"C:\Users\Disha\Desktop\GEN AI\data")
RAW_DIR = BASE_DIR / "raw shapes" / "train"
SEQUENCES_DIR = BASE_DIR / "sequences"
SEQUENCES_DIR.mkdir(parents=True, exist_ok=True)
 
# ── Sequence generation parameters ───────────────────────────
NUM_SEQUENCES   = 100          # total sequences to generate
NUM_FRAMES      = 7            # frames per sequence
IMG_SIZE        = (128, 128)   # (W, H) for every frame
VALID_RATIO     = 0.50         # 50 % valid / 50 % invalid
 
SHAPE_TYPES     = ["circle", "square", "triangle"]
 
VALID_TRANSFORMS   = ["smooth_rotation", "uniform_speed"]
INVALID_TRANSFORMS = ["angle_jump", "axis_change",
                       "shape_morph", "distortion", "frame_discontinuity"]
 
print("✅ Configuration loaded.")
print(f"   Output dir : {SEQUENCES_DIR}")
print(f"   Sequences  : {NUM_SEQUENCES}  |  Frames : {NUM_FRAMES}")
print(f"   Valid/Invalid split : {int(VALID_RATIO*100)} / {int((1-VALID_RATIO)*100)} %")
 
```

```python
def load_dataset(raw_dir: Path, img_size=(128, 128)) -> list[dict]:
    """
    Load all images from raw_shapes/train/<shape>/ folders.
    Returns a list of dicts: {image, label, path}
    """
    dataset = []
    for shape in SHAPE_TYPES:
        folder = raw_dir / shape
        if not folder.exists():
            print(f"  ⚠️  Folder not found: {folder}")
            continue
        for fp in folder.iterdir():
            if fp.suffix.lower() not in {".png", ".jpg", ".jpeg"}:
                continue
            img = cv2.imread(str(fp))
            if img is None:
                continue
            img = cv2.resize(img, img_size)
            dataset.append({"image": img, "label": shape, "path": str(fp)})
    print(f"✅ Dataset loaded — {len(dataset)} images across {SHAPE_TYPES}")
    return dataset
 
dataset = load_dataset(RAW_DIR, IMG_SIZE)
```

```python
#Transformation Engine

class TransformationEngine:
    """Applies individual frame-level transformations."""
 
    @staticmethod
    def rotate(image: np.ndarray, angle: float) -> np.ndarray:
        """Rotate image by `angle` degrees around its centre."""
        h, w = image.shape[:2]
        M = cv2.getRotationMatrix2D((w / 2, h / 2), angle, 1.0)
        return cv2.warpAffine(image, M, (w, h),
                              flags=cv2.INTER_LINEAR,
                              borderMode=cv2.BORDER_REFLECT)
 
    @staticmethod
    def shear(image: np.ndarray, shear_x: float = 0.3,
              shear_y: float = 0.2) -> np.ndarray:
        """Apply shear warp (used for INVALID distortion)."""
        h, w = image.shape[:2]
        src = np.float32([[0, 0], [w, 0], [0, h]])
        dst = np.float32([
            [shear_x * w * 0.2, shear_y * h * 0.1],
            [w - shear_x * w * 0.1, shear_y * h * 0.15],
            [shear_x * w * 0.15, h - shear_y * h * 0.1]
        ])
        M = cv2.getAffineTransform(src, dst)
        return cv2.warpAffine(image, M, (w, h),
                              borderMode=cv2.BORDER_REFLECT)
 
    @staticmethod
    def swap_shape(dataset: list[dict],
                   current_label: str) -> np.ndarray:
        """Return a random image of a DIFFERENT shape (shape morph)."""
        others = [d for d in dataset if d["label"] != current_label]
        return deepcopy(random.choice(others)["image"])
 
    @staticmethod
    def add_noise(image: np.ndarray, intensity: int = 60) -> np.ndarray:
        """Add gaussian noise (frame discontinuity simulation)."""
        noise = np.random.randint(-intensity, intensity,
                                  image.shape, dtype=np.int16)
        noisy = np.clip(image.astype(np.int16) + noise, 0, 255)
        return noisy.astype(np.uint8)
 
print("✅ TransformationEngine ready.")
```

```python
#sequence generator
class SequenceGenerator:
    """
    Generates one complete frame sequence + metadata for a given
    base image and transformation plan.
    """
 
    def __init__(self, dataset: list[dict],
                 num_frames: int = NUM_FRAMES):
        self.dataset    = dataset
        self.num_frames = num_frames
        self.engine     = TransformationEngine()
 
    # ── PUBLIC API ────────────────────────────────────────────
    def generate(self, base_item: dict,
                 transform_type: str) -> dict:
        """
        Returns:
          frames   : list of np.ndarray (len == num_frames)
          metadata : dict
        """
        label = "VALID" if transform_type in VALID_TRANSFORMS else "INVALID"
 
        frames, angles = self._dispatch(base_item, transform_type)
 
        metadata = {
            "shape"         : base_item["label"],
            "transformation": transform_type,
            "angles"        : [round(float(a), 2) for a in angles],
            "label"         : label,
            "num_frames"    : len(frames),
            "seed"          : SEED
        }
        return {"frames": frames, "metadata": metadata}
 
    # ── DISPATCH ──────────────────────────────────────────────
    def _dispatch(self, base_item, transform_type):
        dispatch = {
            "smooth_rotation"   : self._smooth_rotation,
            "uniform_speed"     : self._uniform_speed,
            "angle_jump"        : self._angle_jump,
            "axis_change"       : self._axis_change,
            "shape_morph"       : self._shape_morph,
            "distortion"        : self._distortion,
            "frame_discontinuity": self._frame_discontinuity,
        }
        fn = dispatch.get(transform_type)
        if fn is None:
            raise ValueError(f"Unknown transform: {transform_type}")
        return fn(base_item)
 
    # ── VALID TRANSFORMATIONS ─────────────────────────────────
    def _smooth_rotation(self, item):
        """0° → constant increment, no jumps."""
        start  = random.uniform(0, 30)
        step   = random.uniform(8, 15)
        angles = [start + i * step for i in range(self.num_frames)]
        frames = [self.engine.rotate(item["image"].copy(), a) for a in angles]
        return frames, angles
 
    def _uniform_speed(self, item):
        """Strict constant angular velocity."""
        start  = 0.0
        step   = 10.0          # exactly 10° per frame
        angles = [start + i * step for i in range(self.num_frames)]
        frames = [self.engine.rotate(item["image"].copy(), a) for a in angles]
        return frames, angles
 
    # ── INVALID TRANSFORMATIONS ───────────────────────────────
    def _angle_jump(self, item):
        """Sudden large angle jump at a random frame."""
        step        = random.uniform(8, 15)
        jump_frame  = random.randint(2, self.num_frames - 2)
        jump_amount = random.uniform(100, 160)
        angles = []
        for i in range(self.num_frames):
            if i < jump_frame:
                angles.append(i * step)
            elif i == jump_frame:
                angles.append(angles[-1] + jump_amount)
            else:
                angles.append(angles[-1] + step)
        frames = [self.engine.rotate(item["image"].copy(), a) for a in angles]
        return frames, angles
 
    def _axis_change(self, item):
        """Rotation direction randomly reverses mid-sequence."""
        step      = random.uniform(8, 15)
        flip_at   = random.randint(2, self.num_frames - 2)
        angles, direction = [0.0], 1
        for i in range(1, self.num_frames):
            if i == flip_at:
                direction *= -1              # reverse direction
            angles.append(angles[-1] + direction * step)
        frames = [self.engine.rotate(item["image"].copy(), a) for a in angles]
        return frames, angles
 
    def _shape_morph(self, item):
        """Shape changes to a different shape mid-sequence."""
        morph_at = random.randint(2, self.num_frames - 2)
        step     = random.uniform(8, 15)
        angles   = [i * step for i in range(self.num_frames)]
        frames   = []
        for i, a in enumerate(angles):
            if i < morph_at:
                src = item["image"].copy()
            else:                             # swap to different shape
                src = self.engine.swap_shape(self.dataset, item["label"])
            frames.append(self.engine.rotate(src, a))
        return frames, angles
 
    def _distortion(self, item):
        """Progressive shear distortion increases each frame."""
        step   = random.uniform(8, 15)
        angles = [i * step for i in range(self.num_frames)]
        frames = []
        for i, a in enumerate(angles):
            rotated = self.engine.rotate(item["image"].copy(), a)
            if i >= self.num_frames // 2:    # distort second half
                sx = random.uniform(0.2, 0.5)
                sy = random.uniform(0.1, 0.3)
                rotated = self.engine.shear(rotated, sx, sy)
            frames.append(rotated)
        return frames, angles
 
    def _frame_discontinuity(self, item):
        """Random frame(s) replaced with heavy noise."""
        step         = random.uniform(8, 15)
        angles       = [i * step for i in range(self.num_frames)]
        bad_frames   = random.sample(range(self.num_frames),
                                     k=random.randint(1, 2))
        frames = []
        for i, a in enumerate(angles):
            if i in bad_frames:
                frames.append(self.engine.add_noise(item["image"].copy()))
            else:
                frames.append(self.engine.rotate(item["image"].copy(), a))
        return frames, angles
 
print("✅ SequenceGenerator ready.")
```

```python
#saving utilities
def save_sequence(seq_id: int, sequence: dict,
                  out_dir: Path) -> Path:
    """
    Save frames + metadata.json for one sequence.
    Returns the sequence folder path.
    """
    folder = out_dir / f"sequence_{seq_id:04d}"
    folder.mkdir(parents=True, exist_ok=True)
 
    # Save frames
    for idx, frame in enumerate(sequence["frames"]):
        cv2.imwrite(str(folder / f"frame_{idx}.png"), frame)
 
    # Save metadata
    with open(folder / "metadata.json", "w") as f:
        json.dump(sequence["metadata"], f, indent=2)
 
    return folder
 
 
def save_dataset_summary(summary: list[dict], out_dir: Path):
    """Save a top-level summary JSON with all sequence metadata."""
    with open(out_dir / "dataset_summary.json", "w") as f:
        json.dump(summary, f, indent=2)
    print(f"✅ dataset_summary.json saved → {out_dir / 'dataset_summary.json'}")
 
print("✅ Save utilities ready.")
```

```python
#Main Generation Loop
import json

def run_phase2(dataset        : list[dict],
               num_sequences  : int  = NUM_SEQUENCES,
               valid_ratio    : float = VALID_RATIO,
               out_dir        : Path  = SEQUENCES_DIR) -> list[dict]:
    """
    Generate `num_sequences` sequences with a balanced valid/invalid split.
    Returns summary list.
    """
    generator = SequenceGenerator(dataset, num_frames=NUM_FRAMES)
 
    n_valid   = int(num_sequences * valid_ratio)
    n_invalid = num_sequences - n_valid
 
    # Build transform assignment list (shuffled)
    transform_pool = (
        [random.choice(VALID_TRANSFORMS)   for _ in range(n_valid)] +
        [random.choice(INVALID_TRANSFORMS) for _ in range(n_invalid)]
    )
    random.shuffle(transform_pool)
 
    summary       = []
    valid_count   = 0
    invalid_count = 0
 
    print(f"\n🚀 Generating {num_sequences} sequences …\n")
 
    for seq_id, transform_type in enumerate(transform_pool):
        # Pick a random base image
        base_item = random.choice(dataset)
 
        try:
            sequence = generator.generate(base_item, transform_type)
        except Exception as exc:
            print(f"  ⚠️  seq {seq_id:04d} failed ({exc}), skipping.")
            continue
 
        # Save to disk
        folder = save_sequence(seq_id, sequence, out_dir)
 
        # Track stats
        lbl = sequence["metadata"]["label"]
        if lbl == "VALID":
            valid_count += 1
        else:
            invalid_count += 1
 
        summary.append({
            "sequence_id"   : f"sequence_{seq_id:04d}",
            **sequence["metadata"]
        })
 
        # Progress every 10 sequences
        if (seq_id + 1) % 10 == 0 or seq_id == 0:
            print(f"  [{seq_id+1:>4}/{num_sequences}]  "
                  f"valid={valid_count}  invalid={invalid_count}  "
                  f"last={transform_type}")
 
    save_dataset_summary(summary, out_dir)
 
    print(f"\n{'─'*50}")
    print(f"✅ Phase 2 complete!")
    print(f"   Total sequences  : {len(summary)}")
    print(f"   VALID            : {valid_count}")
    print(f"   INVALID          : {invalid_count}")
    print(f"   Output directory : {out_dir.resolve()}")
    return summary
 
# ── Run ───────────────────────────────────────────────────────
summary = run_phase2(dataset)
```

```python
SMOOTH_JUMP_THRESHOLD  = 30.0   # degrees — jump larger than this → NOT smooth
CONSISTENCY_STD_LIMIT  = 25.0   # std-dev limit for "uniform speed" check
DIRECTION_CHANGE_MIN   = 2      # minimum reversals to flag axis_change
 
print("✅ Phase 3 configuration loaded.")
print(f"   Smooth jump threshold : {SMOOTH_JUMP_THRESHOLD}°")
print(f"   Consistency std limit : {CONSISTENCY_STD_LIMIT}°")
```

```python
#sequence loader
def load_sequence_data(seq_folder: Path) -> dict | None:
    """
    Loads:
      - frames      : list of np.ndarray (BGR), ordered by frame index
      - metadata    : dict from metadata.json
      - angles      : list[float] pulled from metadata
      - shape       : str shape label
      - label       : 'VALID' | 'INVALID'
      - transform   : transformation type string
 
    Returns None if folder is incomplete.
    """
    meta_path = seq_folder / "metadata.json"
    if not meta_path.exists():
        return None
 
    with open(meta_path) as f:
        metadata = json.load(f)
 
    # Sort frames numerically: frame_0, frame_1, …
    frame_files = sorted(
        seq_folder.glob("frame_*.png"),
        key=lambda p: int(p.stem.split("_")[1])
    )
    if not frame_files:
        return None
 
    frames = [cv2.imread(str(fp)) for fp in frame_files]
    frames = [f for f in frames if f is not None]   # drop unreadable
 
    return {
        "seq_id"   : seq_folder.name,
        "frames"   : frames,
        "metadata" : metadata,
        "angles"   : metadata.get("angles", []),
        "shape"    : metadata.get("shape", "unknown"),
        "label"    : metadata.get("label", "UNKNOWN"),
        "transform": metadata.get("transformation", "unknown"),
    }
 
print("✅ load_sequence_data() ready.")
```

```python
#feature extraction
class FeatureExtractor:
    """
    Extracts all motion features from a loaded sequence dict.
    All methods are stateless; call extract() for the full feature set.
    """
 
    def __init__(self,
                 smooth_threshold : float = SMOOTH_JUMP_THRESHOLD,
                 std_limit        : float = CONSISTENCY_STD_LIMIT,
                 dir_change_min   : int   = DIRECTION_CHANGE_MIN):
        self.smooth_threshold = smooth_threshold
        self.std_limit        = std_limit
        self.dir_change_min   = dir_change_min
 
    # ── PUBLIC API ────────────────────────────────────────────
    def extract(self, seq_data: dict) -> dict:
        """
        Master method — returns the complete features dict.
        All feature groups are computed here and merged.
        """
        angles = seq_data["angles"]
 
        angle_feats   = self._angle_features(angles)
        smooth_feats  = self._smoothness_features(angle_feats["angle_diffs"])
        shape_feats   = self._shape_consistency(seq_data)
        temporal_feat = self._temporal_consistency(angles,
                                                   seq_data["frames"])
        advanced      = self._advanced_features(angle_feats["angle_diffs"],
                                                angles)
 
        features = {
            # ── core ──────────────────────────────────────────
            **angle_feats,
            **smooth_feats,
            **shape_feats,
            **temporal_feat,
            # ── advanced ──────────────────────────────────────
            **advanced,
            # ── provenance ────────────────────────────────────
            "frame_count"      : len(seq_data["frames"]),
            "shape"            : seq_data["shape"],
            "transformation"   : seq_data["transform"],
            "ground_truth_label": seq_data["label"],
        }
        return features
 
    # ── STEP 3 : ANGLE FEATURES ───────────────────────────────
    def _angle_features(self, angles: list) -> dict:
        """
        Computes per-frame angle differences and summary stats.
        angle_diffs[i] = |angles[i+1] - angles[i]|
        """
        if len(angles) < 2:
            return {
                "angle_diffs" : [],
                "max_jump"    : 0.0,
                "avg_change"  : 0.0,
            }
 
        diffs = [abs(angles[i+1] - angles[i])
                 for i in range(len(angles) - 1)]
 
        return {
            "angle_diffs" : [round(d, 4) for d in diffs],
            "max_jump"    : round(float(np.max(diffs)),  4),
            "avg_change"  : round(float(np.mean(diffs)), 4),
        }
 
    # ── STEP 4 : SMOOTHNESS ───────────────────────────────────
    def _smoothness_features(self, angle_diffs: list) -> dict:
        """
        is_smooth   → True if no single jump exceeds threshold
        std_dev     → variance in frame-to-frame changes
        is_uniform  → std_dev within std_limit (constant speed)
        """
        if not angle_diffs:
            return {"is_smooth": True, "std_dev": 0.0, "is_uniform": True}
 
        arr        = np.array(angle_diffs)
        std_dev    = round(float(np.std(arr)),  4)
        max_j      = float(np.max(arr))
        is_smooth  = bool(max_j   <= self.smooth_threshold)
        is_uniform = bool(std_dev <= self.std_limit)
 
        return {
            "is_smooth"  : is_smooth,
            "std_dev"    : std_dev,
            "is_uniform" : is_uniform,
        }
 
    # ── STEP 5 : SHAPE CONSISTENCY ────────────────────────────
    def _shape_consistency(self, seq_data: dict) -> dict:
        """
        Primary check  : transformation type implies shape change?
        Secondary check: contour count stability across frames.
 
        shape_consistent = True when shape stays the same across frames.
        contour_std      = std-dev of contour counts (low → stable shape).
        """
        transform = seq_data["transform"]
 
        # If transformation is shape_morph → definitionally inconsistent
        if transform == "shape_morph":
            return {
                "shape_consistent"     : False,
                "contour_std"          : None,
                "contour_count_per_frame": [],
            }
 
        # Lightweight contour-count analysis
        contour_counts = []
        for frame in seq_data["frames"]:
            gray    = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
            blurred = cv2.GaussianBlur(gray, (5, 5), 0)
            _, thr  = cv2.threshold(blurred, 30, 255, cv2.THRESH_BINARY)
            cnts, _ = cv2.findContours(thr, cv2.RETR_EXTERNAL,
                                       cv2.CHAIN_APPROX_SIMPLE)
            # Count only significant contours (area > 50 px)
            sig = [c for c in cnts if cv2.contourArea(c) > 50]
            contour_counts.append(len(sig))
 
        if len(contour_counts) < 2:
            c_std = 0.0
        else:
            c_std = round(float(np.std(contour_counts)), 4)
 
        # Consistent if contour count barely changes (tolerance: std < 1.5)
        shape_consistent = bool(c_std < 1.5)
 
        return {
            "shape_consistent"      : shape_consistent,
            "contour_std"           : c_std,
            "contour_count_per_frame": contour_counts,
        }
 
    # ── STEP 6 : TEMPORAL CONSISTENCY ────────────────────────
    def _temporal_consistency(self, angles: list,
                               frames: list) -> dict:
        """
        frames_complete : do we have the expected number of frames?
        has_large_gap   : any diff > smooth_threshold?
        expected_frames : frame count stored in metadata
        """
        expected        = len(angles)      # angles list defines expected count
        actual          = len(frames)
        frames_complete = bool(actual >= expected)
 
        diffs = ([abs(angles[i+1] - angles[i])
                  for i in range(len(angles) - 1)]
                 if len(angles) >= 2 else [])
 
        has_large_gap = bool(any(d > self.smooth_threshold for d in diffs))
 
        return {
            "frames_complete": frames_complete,
            "expected_frames": expected,
            "actual_frames"  : actual,
            "has_large_gap"  : has_large_gap,
        }
 
    # ── ADVANCED FEATURES ─────────────────────────────────────
    def _advanced_features(self, angle_diffs: list,
                            angles: list) -> dict:
        """
        direction_changes : number of times rotation reverses sign
        monotonic         : rotation always in one direction
        acceleration      : mean of second differences (jerk)
        """
        if len(angles) < 2:
            return {
                "direction_changes": 0,
                "monotonic"        : True,
                "acceleration"     : 0.0,
            }
 
        raw_diffs  = [angles[i+1] - angles[i]
                      for i in range(len(angles) - 1)]
        signs      = [np.sign(d) for d in raw_diffs if d != 0]
        dir_changes= sum(1 for i in range(1, len(signs))
                         if signs[i] != signs[i-1])
        monotonic  = bool(dir_changes == 0)
 
        # Second differences (acceleration / jerk)
        if len(angle_diffs) >= 2:
            second_diffs = [abs(angle_diffs[i+1] - angle_diffs[i])
                            for i in range(len(angle_diffs) - 1)]
            accel = round(float(np.mean(second_diffs)), 4)
        else:
            accel = 0.0
 
        return {
            "direction_changes": int(dir_changes),
            "monotonic"        : monotonic,
            "acceleration"     : accel,
        }
 
print("✅ FeatureExtractor ready.")
```

```python
#save features
def save_features(seq_folder: Path, features: dict) -> Path:
    """Write features.json alongside frames + metadata.json."""
    out_path = seq_folder / "features.json"
 
    # Convert numpy booleans / ints to native Python types for JSON
    def _to_native(obj):
        if isinstance(obj, (np.bool_,)):
            return bool(obj)
        if isinstance(obj, (np.integer,)):
            return int(obj)
        if isinstance(obj, (np.floating,)):
            return float(obj)
        if isinstance(obj, np.ndarray):
            return obj.tolist()
        return obj
 
    clean = {k: _to_native(v) for k, v in features.items()}
 
    with open(out_path, "w") as f:
        json.dump(clean, f, indent=2)
    return out_path
 
print("✅ save_features() ready.")
```

```python
#main extraction loop
def run_phase3(sequences_dir : Path        = SEQUENCES_DIR,
               summary_list  : list[dict]  = None) -> list[dict]:
    """
    Iterate over every sequence folder, extract features, save features.json.
    Returns list of feature dicts (one per sequence).
    """
    extractor     = FeatureExtractor()
    all_features  = []
    errors        = []
 
    # Determine which folders to process
    if summary_list:
        folders = [sequences_dir / s["sequence_id"] for s in summary_list]
    else:
        folders = sorted(sequences_dir.glob("sequence_*"))
 
    total = len(folders)
    print(f"\n🚀 Extracting features for {total} sequences …\n")
 
    for idx, folder in enumerate(folders):
        seq_data = load_sequence_data(folder)
        if seq_data is None:
            errors.append(str(folder))
            continue
 
        try:
            features = extractor.extract(seq_data)
        except Exception as exc:
            errors.append(f"{folder.name}: {exc}")
            continue
 
        save_features(folder, features)
        all_features.append({"seq_id": folder.name, **features})
 
        if (idx + 1) % 10 == 0 or idx == 0:
            print(f"  [{idx+1:>4}/{total}]  {folder.name}  "
                  f"label={features['ground_truth_label']}  "
                  f"smooth={features['is_smooth']}  "
                  f"max_jump={features['max_jump']}°")
 
    # Save master features summary
    master_path = sequences_dir / "features_summary.json"
    with open(master_path, "w") as f:
        json.dump(all_features, f, indent=2,
                  default=lambda o: (bool(o) if isinstance(o, np.bool_)
                                     else int(o) if isinstance(o, np.integer)
                                     else float(o)))
 
    print(f"\n{'─'*55}")
    print(f"✅ Phase 3 complete!")
    print(f"   Sequences processed : {len(all_features)}")
    print(f"   Errors / skipped    : {len(errors)}")
    print(f"   Master summary      : {master_path.resolve()}")
    if errors:
        print(f"   ⚠️  Failed folders  : {errors[:5]} …")
    return all_features
 
# ── Run ───────────────────────────────────────────────────────
all_features = run_phase3(SEQUENCES_DIR, summary)
```

```python
RULES_CONFIG = {
    # RULE 1 — smooth motion
    "smooth_motion": {
        "enabled"    : True,
        "threshold"  : SMOOTH_JUMP_THRESHOLD,   # from Phase 3 config
        "message"    : "Motion is not smooth — sudden changes detected between frames.",
    },
    # RULE 2 — max angle jump
    "angle_jump": {
        "enabled"    : True,
        "threshold"  : 50.0,     # degrees; jump above this is physically implausible
        "message"    : "Sudden angle jump detected — rotation increment exceeds physical limit.",
    },
    # RULE 3 — shape consistency
    "shape_consistency": {
        "enabled"    : True,
        "message"    : "Shape inconsistency detected — object geometry changed mid-sequence.",
    },
    # RULE 4 — temporal gap
    "temporal_gap": {
        "enabled"    : True,
        "message"    : "Temporal discontinuity detected — large gap between consecutive frames.",
    },
    # RULE 5 — uniform speed (warning-level; still counts as violation)
    "uniform_speed": {
        "enabled"    : True,
        "threshold"  : CONSISTENCY_STD_LIMIT,    # from Phase 3 config
        "message"    : "Non-uniform speed detected — angular velocity is inconsistent.",
    },
    # RULE 6 — direction changes
    "direction_change": {
        "enabled"    : True,
        "threshold"  : DIRECTION_CHANGE_MIN,     # from Phase 3 config
        "message"    : "Rotation axis change detected — direction reversed unexpectedly.",
    },
    # RULE 7 — frame completeness
    "frame_completeness": {
        "enabled"    : True,
        "message"    : "Incomplete sequence — actual frame count is less than expected.",
    },
    # RULE 8 — high acceleration / jerk
    "high_acceleration": {
        "enabled"    : True,
        "threshold"  : 20.0,     # mean second-difference in degrees
        "message"    : "High angular acceleration detected — motion is physically jerky.",
    },
}
 
print("✅ Phase 4 rules configuration loaded.")
print(f"   Active rules : "
      f"{sum(1 for r in RULES_CONFIG.values() if r['enabled'])} / {len(RULES_CONFIG)}")
for name, cfg in RULES_CONFIG.items():
    status = "ON " if cfg["enabled"] else "OFF"
    thr    = f"  threshold={cfg.get('threshold')}" if "threshold" in cfg else ""
    print(f"   [{status}] {name:<22}{thr}")
```

```python
#feature loader
def load_features(seq_folder: Path) -> dict | None:
    """
    Load features.json for one sequence.
    Returns dict or None if file missing.
    """
    feat_path = seq_folder / "features.json"
    if not feat_path.exists():
        return None
    with open(feat_path) as f:
        return json.load(f)
 
print("✅ load_features() ready.")
```

```python
#rule engine
class RuleEngine:
    """
    Applies all enabled rules to a features dict.
    Returns a list of violation strings; empty list → VALID.
    """
 
    def __init__(self, config: dict = RULES_CONFIG):
        self.config = config
 
    # ── PUBLIC API ────────────────────────────────────────────
    def evaluate(self, features: dict) -> list[str]:
        """
        Run every enabled rule against `features`.
        Returns violations list.
        """
        violations = []
 
        rule_methods = [
            ("smooth_motion",      self._rule_smooth_motion),
            ("angle_jump",         self._rule_angle_jump),
            ("shape_consistency",  self._rule_shape_consistency),
            ("temporal_gap",       self._rule_temporal_gap),
            ("uniform_speed",      self._rule_uniform_speed),
            ("direction_change",   self._rule_direction_change),
            ("frame_completeness", self._rule_frame_completeness),
            ("high_acceleration",  self._rule_high_acceleration),
        ]
 
        for rule_name, method in rule_methods:
            cfg = self.config.get(rule_name, {})
            if not cfg.get("enabled", False):
                continue
            violated, detail = method(features, cfg)
            if violated:
                msg = cfg["message"]
                if detail:
                    msg = f"{msg}  [{detail}]"
                violations.append(msg)
 
        return violations
 
    # ── RULE 1: Smooth Motion ─────────────────────────────────
    @staticmethod
    def _rule_smooth_motion(features: dict, cfg: dict):
        """
        Violated if is_smooth is explicitly False.
        """
        is_smooth = features.get("is_smooth", True)
        if not is_smooth:
            max_j = features.get("max_jump", 0)
            return True, f"max_jump={max_j}°"
        return False, ""
 
    # ── RULE 2: Angle Jump ────────────────────────────────────
    @staticmethod
    def _rule_angle_jump(features: dict, cfg: dict):
        """
        Violated if max_jump exceeds the angle_jump threshold.
        Threshold is independent of SMOOTH_JUMP_THRESHOLD so
        both rules can fire with different sensitivities.
        """
        max_jump  = features.get("max_jump", 0.0)
        threshold = cfg.get("threshold", 50.0)
        if max_jump > threshold:
            return True, f"max_jump={max_jump}° > threshold={threshold}°"
        return False, ""
 
    # ── RULE 3: Shape Consistency ─────────────────────────────
    @staticmethod
    def _rule_shape_consistency(features: dict, cfg: dict):
        """
        Violated if shape_consistent is False.
        """
        consistent = features.get("shape_consistent", True)
        if not consistent:
            c_std = features.get("contour_std", "N/A")
            return True, f"contour_std={c_std}"
        return False, ""
 
    # ── RULE 4: Temporal Gap ──────────────────────────────────
    @staticmethod
    def _rule_temporal_gap(features: dict, cfg: dict):
        """
        Violated if has_large_gap is True.
        """
        has_gap = features.get("has_large_gap", False)
        if has_gap:
            avg = features.get("avg_change", "N/A")
            return True, f"avg_change={avg}°"
        return False, ""
 
    # ── RULE 5: Uniform Speed ─────────────────────────────────
    @staticmethod
    def _rule_uniform_speed(features: dict, cfg: dict):
        """
        Violated if is_uniform is False (std_dev too high).
        """
        is_uniform = features.get("is_uniform", True)
        if not is_uniform:
            std = features.get("std_dev", "N/A")
            threshold = cfg.get("threshold", CONSISTENCY_STD_LIMIT)
            return True, f"std_dev={std} > limit={threshold}"
        return False, ""
 
    # ── RULE 6: Direction Changes ─────────────────────────────
    @staticmethod
    def _rule_direction_change(features: dict, cfg: dict):
        """
        Violated if direction_changes exceeds threshold.
        """
        dir_changes = features.get("direction_changes", 0)
        threshold   = cfg.get("threshold", DIRECTION_CHANGE_MIN)
        if dir_changes >= threshold:
            return True, f"direction_changes={dir_changes} >= threshold={threshold}"
        return False, ""
 
    # ── RULE 7: Frame Completeness ────────────────────────────
    @staticmethod
    def _rule_frame_completeness(features: dict, cfg: dict):
        """
        Violated if actual_frames < expected_frames.
        """
        complete  = features.get("frames_complete", True)
        if not complete:
            expected = features.get("expected_frames", "?")
            actual   = features.get("actual_frames",   "?")
            return True, f"expected={expected}  actual={actual}"
        return False, ""
 
    # ── RULE 8: High Acceleration ─────────────────────────────
    @staticmethod
    def _rule_high_acceleration(features: dict, cfg: dict):
        """
        Violated if mean angular acceleration (jerk) is too high.
        """
        accel     = features.get("acceleration", 0.0)
        threshold = cfg.get("threshold", 20.0)
        if accel > threshold:
            return True, f"acceleration={accel}° > threshold={threshold}°"
        return False, ""
 
print("✅ RuleEngine ready.")
print(f"   Rules registered : {len([m for m in dir(RuleEngine) if m.startswith('_rule_')])}")
 
```

```python
#decision logic
def make_decision(violations: list[str]) -> str:
    """
    Core decision:  empty violations → VALID,  else → INVALID.
    """
    return "VALID" if len(violations) == 0 else "INVALID"
 
 
def save_validation(seq_folder: Path,
                    final_label : str,
                    violations  : list[str],
                    features    : dict) -> Path:
    """
    Write validation.json to the sequence folder.
    Includes ground-truth label for later accuracy scoring.
    """
    result = {
        "final_label"        : final_label,
        "violations"         : violations,
        "violation_count"    : len(violations),
        "ground_truth_label" : features.get("ground_truth_label", "UNKNOWN"),
        "correct_prediction" : (
            final_label == features.get("ground_truth_label", "")
        ),
        "rules_fired"        : [
            v.split("—")[0].strip() for v in violations   # short rule name
        ],
    }
    out_path = seq_folder / "validation.json"
    with open(out_path, "w") as f:
        json.dump(result, f, indent=2)
    return out_path
 
print("✅ make_decision() and save_validation() ready.")
 
```

```python
#main validation loop
def run_phase4(sequences_dir : Path       = SEQUENCES_DIR,
               summary_list  : list[dict] = None) -> list[dict]:
    """
    For every sequence:
      1. Load features.json
      2. Evaluate all rules
      3. Make VALID / INVALID decision
      4. Save validation.json
    Returns a list of validation result dicts.
    """
    engine      = RuleEngine(RULES_CONFIG)
    results     = []
    errors      = []
 
    if summary_list:
        folders = [sequences_dir / s["sequence_id"] for s in summary_list]
    else:
        folders = sorted(sequences_dir.glob("sequence_*"))
 
    total = len(folders)
    print(f"\n🚀 Validating {total} sequences …\n")
 
    correct = 0
    for idx, folder in enumerate(folders):
 
        features = load_features(folder)
        if features is None:
            errors.append(str(folder))
            continue
 
        violations  = engine.evaluate(features)
        final_label = make_decision(violations)
        save_validation(folder, final_label, violations, features)
 
        gt      = features.get("ground_truth_label", "?")
        is_ok   = "✓" if final_label == gt else "✗"
        if final_label == gt:
            correct += 1
 
        results.append({
            "seq_id"             : folder.name,
            "final_label"        : final_label,
            "ground_truth_label" : gt,
            "correct_prediction" : final_label == gt,
            "violation_count"    : len(violations),
            "violations"         : violations,
        })
 
        if (idx + 1) % 10 == 0 or idx == 0:
            acc = correct / (idx + 1) * 100
            print(f"  [{idx+1:>4}/{total}]  {folder.name}  "
                  f"pred={final_label:<8} gt={gt:<8} {is_ok}  "
                  f"violations={len(violations)}  "
                  f"running_acc={acc:.1f}%")
 
    # ── Master validation summary ─────────────────────────────
    master_path = sequences_dir / "validation_summary.json"
    with open(master_path, "w") as f:
        json.dump(results, f, indent=2)
 
    accuracy = correct / len(results) * 100 if results else 0
    print(f"\n{'─'*60}")
    print(f"✅ Phase 4 complete!")
    print(f"   Sequences validated : {len(results)}")
    print(f"   Errors / skipped    : {len(errors)}")
    print(f"   Correct predictions : {correct} / {len(results)}")
    print(f"   Rule-Engine Accuracy: {accuracy:.2f}%")
    print(f"   Summary saved       : {master_path.resolve()}")
    return results
 
# ── Run ───────────────────────────────────────────────────────
validation_results = run_phase4(SEQUENCES_DIR, summary)
 
```

```python
#visualizing decision
def visualize_validation(seq_folder: Path):
    """
    Side-by-side frame strip with validation overlay.
    Green border → VALID prediction  |  Red border → INVALID prediction
    """
    feat_path = seq_folder / "features.json"
    val_path  = seq_folder / "validation.json"
    if not feat_path.exists() or not val_path.exists():
        print("Missing features.json or validation.json")
        return
 
    with open(feat_path) as f: feat = json.load(f)
    with open(val_path)  as f: val  = json.load(f)
 
    frame_files = sorted(
        seq_folder.glob("frame_*.png"),
        key=lambda p: int(p.stem.split("_")[1])
    )
    frames = [cv2.cvtColor(cv2.imread(str(fp)), cv2.COLOR_BGR2RGB)
              for fp in frame_files]
 
    n   = len(frames)
    fig, axes = plt.subplots(1, n, figsize=(3 * n, 3.8))
    if n == 1:
        axes = [axes]
 
    pred_color = "#2ecc71" if val["final_label"] == "VALID" else "#e74c3c"
 
    for i, (ax, frame) in enumerate(zip(axes, frames)):
        ax.imshow(frame)
        angle_str = (f"{feat['angle_diffs'][i-1]:.1f}°Δ"
                     if i > 0 and i - 1 < len(feat.get("angle_diffs", []))
                     else "—")
        ax.set_title(f"f{i}\n{angle_str}", fontsize=8)
        ax.axis("off")
        for spine in ax.spines.values():
            spine.set_edgecolor(pred_color)
            spine.set_linewidth(3)
 
    gt    = val["ground_truth_label"]
    pred  = val["final_label"]
    match = "✓ Correct" if val["correct_prediction"] else "✗ Wrong"
    viols = "\n".join(f"  • {v.split('—')[0].strip()}"
                      for v in val["violations"]) or "  None"
 
    title = (f"Shape: {feat.get('shape','?').upper()}  |  "
             f"Transform: {feat.get('transformation','?')}  |  "
             f"GT: {gt}  |  Pred: {pred}  |  {match}")
    fig.suptitle(title, fontsize=10, color=pred_color,
                 fontweight="bold", y=1.0)
    fig.text(0.5, -0.04,
             f"Violations:\n{viols}",
             ha="center", fontsize=8.5, color="#555555")
    plt.tight_layout()
    plt.show()
 
# ── Show one correctly validated INVALID sequence ─────────────
for r in validation_results:
    if r["final_label"] == "INVALID" and r["correct_prediction"]:
        visualize_validation(SEQUENCES_DIR / r["seq_id"])
        break
 
# ── Show one misclassified sequence (if any) ──────────────────
for r in validation_results:
    if not r["correct_prediction"]:
        print(f"\n⚠️  Misclassified example:")
        visualize_validation(SEQUENCES_DIR / r["seq_id"])
        break
```

```python
#local embedding engine
class EmbeddingEngine:
    """
    Wraps SentenceTransformer for:
      - Single text encoding
      - Batch encoding
      - L2 normalisation (cosine sim = inner product with FAISS)
    Model is loaded once at construction.
    """
 
    def __init__(self, model_name: str = EMBEDDING_MODEL):
        print(f"   Loading embedding model: {model_name} …")
        self.model      = SentenceTransformer(model_name)
        self.model_name = model_name
        self.dim        = EMBEDDING_DIM
        print(f"   ✅ Model loaded  |  dim={self.dim}")
 
    # ── PUBLIC API ────────────────────────────────────────────
    def encode(self, texts: list[str] | str,
               batch_size: int = 32) -> np.ndarray:
        """
        Encode one or more texts.
        Returns float32 numpy array of shape (N, dim), L2-normalised.
        """
        if isinstance(texts, str):
            texts = [texts]
 
        # Clean text — replace newlines that confuse tokeniser
        cleaned = [t.replace("\n", " ").strip() for t in texts]
 
        vectors = self.model.encode(
            cleaned,
            batch_size       = batch_size,
            convert_to_numpy = True,
            show_progress_bar= False,
        )
 
        vectors = vectors.astype(np.float32)
        vectors = self._l2_normalize(vectors)
        return vectors
 
    def encode_single(self, text: str) -> np.ndarray:
        """Encode a single query string → shape (1, dim)."""
        return self.encode([text])
 
    # ── PRIVATE ───────────────────────────────────────────────
    @staticmethod
    def _l2_normalize(vectors: np.ndarray) -> np.ndarray:
        """
        L2-normalise each row so cosine similarity == inner product.
        CRITICAL: required for IndexFlatIP (FAISS inner product index).
        """
        norms = np.linalg.norm(vectors, axis=1, keepdims=True)
        norms = np.where(norms == 0, 1e-10, norms)   # avoid div-by-zero
        return vectors / norms
 
# ── Instantiate once — reused by VectorDB and retrieval ───────
embedding_engine = EmbeddingEngine(EMBEDDING_MODEL)
```

```python
#vector database 
class VectorDB:
    """
    FAISS-backed vector store for rule embeddings.
 
    build()  → encode knowledge base + add to index
    search() → query → top-k rules with scores
    save()   → persist index to disk
    load()   → restore from disk
    """
 
    def __init__(self, dim: int = EMBEDDING_DIM):
        self.dim     = dim
        self.index   = faiss.IndexFlatIP(dim)   # inner product = cosine
        self.entries : list[dict] = []           # parallel rule metadata
 
    # ── BUILD ────────────────────────────────────────────────
    def build(self, knowledge_base: list[dict],
              engine: EmbeddingEngine) -> None:
        """
        Encode all rules and add to FAISS index.
        Clears any existing index first.
        """
        self.index   = faiss.IndexFlatIP(self.dim)
        self.entries = []
 
        rule_texts = [r["rule"] for r in knowledge_base]
 
        print(f"   Encoding {len(rule_texts)} rules …")
        vectors = engine.encode(rule_texts)        # (N, 384) float32
 
        self.index.add(vectors)
        self.entries = knowledge_base.copy()
 
        print(f"   ✅ FAISS index built  |  "
              f"total vectors = {self.index.ntotal}")
 
    # ── SEARCH ───────────────────────────────────────────────
    def search(self, query: str,
               engine: EmbeddingEngine,
               top_k: int = TOP_K_RULES) -> list[dict]:
        """
        Encode query and retrieve top-k matching rules.
 
        Returns list of dicts:
          { id, category, rule, related_violation, score }
        """
        if self.index.ntotal == 0:
            raise RuntimeError("Vector DB is empty. Call build() first.")
 
        q_vec = engine.encode_single(query)        # (1, 384)
        scores, indices = self.index.search(q_vec, top_k)
 
        results = []
        for score, idx in zip(scores[0], indices[0]):
            if idx < 0 or idx >= len(self.entries):
                continue
            entry = self.entries[idx].copy()
            entry["score"] = round(float(score), 4)
            results.append(entry)
 
        return results
 
    # ── PERSIST ──────────────────────────────────────────────
    def save(self, path: Path) -> None:
        faiss.write_index(self.index, str(path))
        print(f"   ✅ FAISS index saved → {path}")
 
    def load(self, path: Path,
             knowledge_base: list[dict]) -> None:
        self.index   = faiss.read_index(str(path))
        self.entries = knowledge_base.copy()
        print(f"   ✅ FAISS index loaded → {self.index.ntotal} vectors")
 
# ── Build the vector database ─────────────────────────────────
print("🔨 Building vector database …")
vector_db = VectorDB(EMBEDDING_DIM)
vector_db.build(KNOWLEDGE_BASE, embedding_engine)
vector_db.save(RAG_INDEX_PATH)
```

```python
#retrival engine
class RetrievalEngine:
    """
    High-level retrieval interface used by Phase 6.
 
    retrieve_for_violations() → top-k rules for a violation list
    build_context()           → formatted string for LLM prompt
    """
 
    def __init__(self, vector_db : VectorDB,
                       engine    : EmbeddingEngine,
                       top_k     : int = TOP_K_RULES):
        self.vector_db = vector_db
        self.engine    = engine
        self.top_k     = top_k
 
    # ── PUBLIC API ────────────────────────────────────────────
    def retrieve_for_violations(self, violations: list[str]) -> list[dict]:
        """
        Merge violations into a single query and retrieve top-k rules.
        """
        query   = self._violations_to_query(violations)
        results = self.vector_db.search(query, self.engine, self.top_k)
        return results
 
    def build_context(self, violations : list[str],
                            retrieved  : list[dict]) -> str:
        """
        Build the context string that Phase 6 will inject into the
        LLM prompt.  Format:
 
          DETECTED VIOLATIONS:
          - violation 1
          ...
 
          RELEVANT PHYSICS RULES:
          [1] rule text  (category, score)
          ...
        """
        lines = ["DETECTED VIOLATIONS:"]
        if violations:
            for v in violations:
                lines.append(f"  - {v}")
        else:
            lines.append("  None")
 
        lines.append("")
        lines.append("RELEVANT PHYSICS RULES:")
        for i, r in enumerate(retrieved, 1):
            lines.append(
                f"  [{i}] {r['rule']}"
                f"  (category: {r['category']}, "
                f"similarity: {r['score']:.3f})"
            )
 
        return "\n".join(lines)
 
    # ── PRIVATE ───────────────────────────────────────────────
    @staticmethod
    def _violations_to_query(violations: list[str]) -> str:
        """
        Join violation messages into a single query sentence.
        Strips inline details (content inside [ ]) for cleaner retrieval.
        """
        cleaned = []
        for v in violations:
            # Remove bracketed details added by RuleEngine
            main = v.split("[")[0].strip().rstrip(".")
            cleaned.append(main)
        return ". ".join(cleaned) + "." if cleaned else "motion validation query"
 
# ── Instantiate retrieval engine ──────────────────────────────
retrieval_engine = RetrievalEngine(vector_db, embedding_engine, TOP_K_RULES)
 
print("✅ RetrievalEngine ready.")
```

```python
# Main RAG Loop over Phase 4 Results
# For every sequence, retrieve relevant rules and build context.
# Saves rag_output.json per sequence.
def run_phase5(sequences_dir     : Path       = SEQUENCES_DIR,
               validation_res    : list[dict] = None) -> list[dict]:
    """
    For each sequence:
      1. Load validation.json (violations from Phase 4)
      2. Retrieve top-k rules via RAG
      3. Build context string
      4. Save rag_output.json
    Returns list of RAG output dicts.
    """
    rag_outputs = []
    errors      = []
 
    if validation_res:
        folders = [sequences_dir / r["seq_id"] for r in validation_res]
    else:
        folders = sorted(sequences_dir.glob("sequence_*"))
 
    total = len(folders)
    print(f"\n🚀 Running RAG retrieval for {total} sequences …\n")
 
    for idx, folder in enumerate(folders):
        val_path = folder / "validation.json"
        if not val_path.exists():
            errors.append(str(folder))
            continue
 
        with open(val_path) as f:
            val_data = json.load(f)
 
        violations = val_data.get("violations", [])
        final_lbl  = val_data.get("final_label", "UNKNOWN")
 
        # ── Retrieve ──────────────────────────────────────────
        retrieved = retrieval_engine.retrieve_for_violations(violations)
        context   = retrieval_engine.build_context(violations, retrieved)
 
        # ── Save rag_output.json ──────────────────────────────
        rag_out = {
            "seq_id"          : folder.name,
            "final_label"     : final_lbl,
            "violations"      : violations,
            "retrieved_rules" : retrieved,
            "context"         : context,
        }
        with open(folder / "rag_output.json", "w") as f:
            json.dump(rag_out, f, indent=2)
 
        rag_outputs.append(rag_out)
 
        if (idx + 1) % 10 == 0 or idx == 0:
            top_score = retrieved[0]["score"] if retrieved else 0.0
            print(f"  [{idx+1:>4}/{total}]  {folder.name}  "
                  f"label={final_lbl:<8}  "
                  f"violations={len(violations)}  "
                  f"top_score={top_score:.3f}")
 
    # ── Master RAG summary ────────────────────────────────────
    master_path = sequences_dir / "rag_summary.json"
    slim = [{k: v for k, v in o.items() if k != "context"}
            for o in rag_outputs]           # omit long context strings
    with open(master_path, "w") as f:
        json.dump(slim, f, indent=2)
 
    print(f"\n{'─'*60}")
    print(f"✅ Phase 5 complete!")
    print(f"   Sequences processed : {len(rag_outputs)}")
    print(f"   Errors / skipped    : {len(errors)}")
    print(f"   Master summary      : {master_path.resolve()}")
    return rag_outputs
 
# ── Run ───────────────────────────────────────────────────────
rag_outputs = run_phase5(SEQUENCES_DIR, validation_results)
```

```python
#prompt engineer
class PromptEngineer:
    """
    Builds structured prompts for the LLM.
 
    Two prompt types:
      build_invalid_prompt() → explanation for INVALID sequences
      build_valid_prompt()   → confirmation for VALID sequences
    """
 
    # ── System persona (sent as system message) ───────────────
    SYSTEM_PROMPT = (
        "You are a physics and geometry expert analysing motion sequences "
        "generated by AI. Your job is to explain whether a sequence of "
        "geometric shape transformations is physically valid or invalid. "
        "Be clear, concise, and avoid unnecessary jargon. "
        "Limit your response to 3-4 sentences."
    )
 
    # ── INVALID template ──────────────────────────────────────
    INVALID_TEMPLATE = """\
Motion Analysis Report
======================
Shape          : {shape}
Transformation : {transformation}
Decision       : INVALID
 
Detected Violations:
{violations_block}
 
Relevant Physics Rules (retrieved from knowledge base):
{rules_block}
 
Task:
Explain clearly and concisely why this motion sequence is physically \
inconsistent. Reference the specific violations and the physics rules \
above. Do not invent new facts. Keep the explanation to 3-4 sentences."""
 
    # ── VALID template ────────────────────────────────────────
    VALID_TEMPLATE = """\
Motion Analysis Report
======================
Shape          : {shape}
Transformation : {transformation}
Decision       : VALID
 
No violations were detected in this sequence.
 
Relevant Physics Principles (for reference):
{rules_block}
 
Task:
Briefly confirm why this motion sequence is physically valid. \
Reference the principles above. Keep the explanation to 2-3 sentences."""
 
    # ── PUBLIC API ────────────────────────────────────────────
    def build(self, rag_out: dict, meta: dict) -> tuple[str, str]:
        """
        Returns (system_prompt, user_prompt) tuple.
 
        rag_out : dict from rag_output.json
        meta    : dict from metadata.json (shape, transformation)
        """
        shape      = meta.get("shape", "unknown").capitalize()
        transform  = meta.get("transformation", "unknown")
        violations = rag_out.get("violations", [])
        retrieved  = rag_out.get("retrieved_rules", [])
 
        v_block = self._format_violations(violations)
        r_block = self._format_rules(retrieved)
 
        if rag_out.get("final_label", "VALID") == "INVALID":
            user_prompt = self.INVALID_TEMPLATE.format(
                shape              = shape,
                transformation     = transform,
                violations_block   = v_block,
                rules_block        = r_block,
            )
        else:
            user_prompt = self.VALID_TEMPLATE.format(
                shape          = shape,
                transformation = transform,
                rules_block    = r_block,
            )
 
        return self.SYSTEM_PROMPT, user_prompt
 
    # ── PRIVATE ───────────────────────────────────────────────
    @staticmethod
    def _format_violations(violations: list[str]) -> str:
        if not violations:
            return "  (none)"
        return "\n".join(f"  • {v.split('[')[0].strip()}"
                         for v in violations)
 
    @staticmethod
    def _format_rules(rules: list[dict]) -> str:
        if not rules:
            return "  (none retrieved)"
        lines = []
        for i, r in enumerate(rules, 1):
            lines.append(
                f"  [{i}] ({r.get('category','?')})  {r['rule']}"
                f"  [similarity: {r.get('score', 0):.3f}]"
            )
        return "\n".join(lines)
 
prompt_engineer = PromptEngineer()
print("✅ PromptEngineer ready.")
 
# ── Quick preview of a prompt ─────────────────────────────────
if rag_outputs:
    _sample = next((o for o in rag_outputs if o["final_label"] == "INVALID"),
                   rag_outputs[0])
    _meta_path = SEQUENCES_DIR / _sample["seq_id"] / "metadata.json"
    _meta = json.load(open(_meta_path)) if _meta_path.exists() else {}
    _sys, _usr = prompt_engineer.build(_sample, _meta)
    print("\n📝 Sample prompt preview (user message):\n")
    print(_usr[:900] + ("\n…[truncated]" if len(_usr) > 900 else ""))
```

```python
#LLM client
class LLMClient:
    """
    Thin wrapper around OpenAI ChatCompletion.
 
    Features:
      - Single call per sequence (efficient)
      - Retry on rate-limit / transient error
      - Fallback explanation if all retries fail
    """
 
    def __init__(self,
                 model       : str   = LLM_MODEL,
                 max_tokens  : int   = LLM_MAX_TOKENS,
                 temperature : float = LLM_TEMPERATURE,
                 retry_limit : int   = LLM_RETRY_LIMIT,
                 retry_delay : int   = LLM_RETRY_DELAY):
        self.model       = model
        self.max_tokens  = max_tokens
        self.temperature = temperature
        self.retry_limit = retry_limit
        self.retry_delay = retry_delay
        self.client      = openai.OpenAI(api_key=OPENAI_API_KEY)
 
    # ── PUBLIC API ────────────────────────────────────────────
    def generate(self, system_prompt: str,
                        user_prompt  : str) -> str:
        """
        Call LLM and return explanation string.
        Retries up to retry_limit times on failure.
        Returns a fallback string if all attempts fail.
        """
        messages = [
            {"role": "system", "content": system_prompt},
            {"role": "user",   "content": user_prompt},
        ]
 
        for attempt in range(1, self.retry_limit + 1):
            try:
                response = self.client.chat.completions.create(
                    model       = self.model,
                    messages    = messages,
                    max_tokens  = self.max_tokens,
                    temperature = self.temperature,
                )
                text = response.choices[0].message.content.strip()
                return text
 
            except openai.RateLimitError:
                if attempt < self.retry_limit:
                    print(f"      ⏳ Rate limit hit — waiting {self.retry_delay}s "
                          f"(attempt {attempt}/{self.retry_limit})")
                    time.sleep(self.retry_delay)
                else:
                    return self._fallback(user_prompt, "rate limit exceeded")
 
            except openai.APIError as exc:
                if attempt < self.retry_limit:
                    time.sleep(self.retry_delay)
                else:
                    return self._fallback(user_prompt, str(exc))
 
            except Exception as exc:
                return self._fallback(user_prompt, str(exc))
 
        return self._fallback(user_prompt, "unknown error")
 
    # ── PRIVATE ───────────────────────────────────────────────
    @staticmethod
    def _fallback(prompt: str, reason: str) -> str:
        """
        Rule-based fallback explanation when LLM is unavailable.
        Parses the prompt to extract key info so the output is still useful.
        """
        is_invalid = "INVALID" in prompt
        if is_invalid:
            return (
                "[FALLBACK — LLM unavailable: " + reason + "] "
                "This motion sequence has been flagged as physically invalid "
                "by the rule engine. One or more detected violations — such as "
                "a sudden angle jump, shape inconsistency, or temporal "
                "discontinuity — contradict the principles of rigid-body motion. "
                "A physically valid sequence requires smooth, continuous rotation "
                "with shape preservation and no abrupt directional changes."
            )
        else:
            return (
                "[FALLBACK — LLM unavailable: " + reason + "] "
                "This motion sequence has been validated as physically consistent. "
                "The rotation is smooth, the shape is preserved across all frames, "
                "and the angular velocity remains uniform — all consistent with "
                "real-world rigid-body dynamics."
            )
 
llm_client = LLMClient()
print("✅ LLMClient ready.")
```

```python
#output builder and saver
def build_explanation_output(rag_out     : dict,
                              meta        : dict,
                              features    : dict,
                              explanation : str) -> dict:
    """
    Assemble the final explanation.json payload.
    Includes full traceability back to features and violations.
    """
    return {
        "seq_id"             : rag_out["seq_id"],
        "shape"              : meta.get("shape", "unknown"),
        "transformation"     : meta.get("transformation", "unknown"),
        "status"             : rag_out["final_label"],
        "ground_truth_label" : features.get("ground_truth_label", "UNKNOWN"),
        "correct_prediction" : (
            rag_out["final_label"] == features.get("ground_truth_label", "")
        ),
        "violation_count"    : len(rag_out.get("violations", [])),
        "violations"         : rag_out.get("violations", []),
        "retrieved_rules"    : [
            {"rule": r["rule"], "category": r["category"], "score": r["score"]}
            for r in rag_out.get("retrieved_rules", [])
        ],
        "explanation"        : explanation,
        "llm_model"          : LLM_MODEL,
    }
 
 
def save_explanation(seq_folder  : Path,
                     output      : dict,
                     also_txt    : bool = True) -> None:
    """
    Save explanation.json (and optionally explanation.txt)
    inside the sequence folder.
    """
    json_path = seq_folder / "explanation.json"
    with open(json_path, "w") as f:
        json.dump(output, f, indent=2)
 
    if also_txt:
        txt_path = seq_folder / "explanation.txt"
        with open(txt_path, "w") as f:
            f.write(f"Sequence  : {output['seq_id']}\n")
            f.write(f"Shape     : {output['shape']}\n")
            f.write(f"Transform : {output['transformation']}\n")
            f.write(f"Status    : {output['status']}\n\n")
            f.write("Violations:\n")
            for v in output["violations"]:
                f.write(f"  • {v}\n")
            f.write("\nExplanation:\n")
            f.write(output["explanation"] + "\n")
 
print("✅ build_explanation_output() and save_explanation() ready.")
 
```

```python
#Main LLM Loop
# Iterates over RAG outputs, calls LLM, saves explanations.
def run_phase6(sequences_dir : Path       = SEQUENCES_DIR,
               rag_out_list  : list[dict] = None,
               max_sequences : int | None = MAX_SEQUENCES_TO_EXPLAIN
               ) -> list[dict]:
    """
    For each sequence (up to max_sequences):
      1. Load rag_output.json  (violations + retrieved rules)
      2. Load metadata.json    (shape, transformation)
      3. Load features.json    (ground truth label)
      4. Build prompt
      5. Call LLM → explanation
      6. Save explanation.json + explanation.txt
    Returns list of explanation output dicts.
    """
    pool = rag_out_list if rag_out_list else []
    if not pool:
        # Fall back: scan folders directly
        folders = sorted(sequences_dir.glob("sequence_*"))
        for fld in folders:
            rp = fld / "rag_output.json"
            if rp.exists():
                with open(rp) as f:
                    pool.append(json.load(f))
 
    if max_sequences is not None:
        pool = pool[:max_sequences]
 
    all_outputs = []
    errors      = []
    total       = len(pool)
 
    print(f"\n🚀 Generating LLM explanations for {total} sequences …\n")
 
    for idx, rag_out in enumerate(pool):
        seq_id  = rag_out["seq_id"]
        folder  = sequences_dir / seq_id
 
        # ── Load supporting files ─────────────────────────────
        meta_path = folder / "metadata.json"
        feat_path = folder / "features.json"
 
        meta     = json.load(open(meta_path)) if meta_path.exists() else {}
        features = json.load(open(feat_path)) if feat_path.exists() else {}
 
        # ── Build prompt ──────────────────────────────────────
        sys_prompt, usr_prompt = prompt_engineer.build(rag_out, meta)
 
        # ── Call LLM ──────────────────────────────────────────
        explanation = llm_client.generate(sys_prompt, usr_prompt)
 
        # ── Assemble + save ───────────────────────────────────
        output = build_explanation_output(rag_out, meta, features, explanation)
        save_explanation(folder, output, also_txt=True)
 
        # Also save to global explanations folder for easy access
        with open(EXPLANATIONS_DIR / f"{seq_id}.json", "w") as f:
            json.dump(output, f, indent=2)
 
        all_outputs.append(output)
        is_fallback = explanation.startswith("[FALLBACK")
        tag = "🔄" if is_fallback else "✅"
 
        if (idx + 1) % 5 == 0 or idx == 0:
            print(f"  {tag} [{idx+1:>4}/{total}]  {seq_id}  "
                  f"status={output['status']:<8}  "
                  f"violations={output['violation_count']}  "
                  f"{'(fallback)' if is_fallback else ''}")
 
    # ── Master explanations summary ───────────────────────────
    master_path = sequences_dir / "explanations_summary.json"
    with open(master_path, "w") as f:
        json.dump(all_outputs, f, indent=2)
 
    fallbacks = sum(1 for o in all_outputs
                    if o["explanation"].startswith("[FALLBACK"))
    print(f"\n{'─'*60}")
    print(f"✅ Phase 6 complete!")
    print(f"   Sequences explained  : {len(all_outputs)}")
    print(f"   LLM calls succeeded  : {len(all_outputs) - fallbacks}")
    print(f"   Fallback (no API)    : {fallbacks}")
    print(f"   Master summary       : {master_path.resolve()}")
    print(f"   Explanations folder  : {EXPLANATIONS_DIR.resolve()}")
    return all_outputs
 
# ── Run ───────────────────────────────────────────────────────
explanation_outputs = run_phase6(SEQUENCES_DIR, rag_outputs,
                                 MAX_SEQUENCES_TO_EXPLAIN)
```

```python
#verification (print sample explanation)
print("=" * 65)
print("📋 PHASE 6 — SAMPLE EXPLANATIONS")
print("=" * 65)
 
def print_explanation(output: dict):
    border = "─" * 65
    status_icon = "✅" if output["status"] == "VALID" else "❌"
    correct_icon = "✓" if output["correct_prediction"] else "✗"
    print(f"\n{border}")
    print(f"{status_icon} Sequence  : {output['seq_id']}")
    print(f"   Shape     : {output['shape']}")
    print(f"   Transform : {output['transformation']}")
    print(f"   Status    : {output['status']}   "
          f"(GT: {output['ground_truth_label']}  {correct_icon})")
    if output["violations"]:
        print("   Violations:")
        for v in output["violations"]:
            print(f"     • {v.split('[')[0].strip()}")
    else:
        print("   Violations: None")
    print(f"\n   💬 Explanation:\n")
    # Wrap explanation text at ~65 chars for readability
    words  = output["explanation"].split()
    line, lines_out = [], []
    for w in words:
        if sum(len(x) + 1 for x in line) + len(w) > 63:
            lines_out.append("   " + " ".join(line))
            line = [w]
        else:
            line.append(w)
    if line:
        lines_out.append("   " + " ".join(line))
    print("\n".join(lines_out))
 
# ── Show 2 INVALID + 1 VALID ─────────────────────────────────
shown_invalid, shown_valid = 0, 0
for out in explanation_outputs:
    if out["status"] == "INVALID" and shown_invalid < 2:
        print_explanation(out)
        shown_invalid += 1
    elif out["status"] == "VALID" and shown_valid < 1:
        print_explanation(out)
        shown_valid += 1
    if shown_invalid >= 2 and shown_valid >= 1:
        break
```

```python
#aggregated stats
edf = pd.DataFrame(explanation_outputs)
 
print("\n" + "=" * 65)
print("📊 PHASE 6 — EXPLANATION STATISTICS")
print("=" * 65)
 
# ── Status distribution ───────────────────────────────────────
print("\n🏷️  Status distribution:")
for lbl, grp in edf.groupby("status"):
    icon = "✅" if lbl == "VALID" else "❌"
    bar  = "█" * len(grp)
    print(f"  {icon} {lbl:<10} {len(grp):>4}  {bar}")
 
# ── Prediction accuracy ───────────────────────────────────────
acc = edf["correct_prediction"].mean() * 100
print(f"\n🎯 Prediction accuracy (Phase 6 subset) : {acc:.1f}%")
 
# ── Explanation length distribution ──────────────────────────
edf["expl_words"] = edf["explanation"].apply(lambda t: len(t.split()))
avg_words = edf["expl_words"].mean()
print(f"\n📝 Explanation length (words):")
print(f"   Mean : {avg_words:.1f}")
print(f"   Min  : {edf['expl_words'].min()}")
print(f"   Max  : {edf['expl_words'].max()}")
 
# ── Fallback rate ─────────────────────────────────────────────
fallback_count = edf["explanation"].str.startswith("[FALLBACK").sum()
print(f"\n🔄 Fallback explanations : {fallback_count} / {len(edf)}")
if fallback_count == len(edf):
    print("   ℹ️  All explanations used rule-based fallback.")
    print("      Set OPENAI_API_KEY to enable live LLM calls.")
 
# ── Sample explanation.json ───────────────────────────────────
if explanation_outputs:
    sample = explanation_outputs[0]
    sample_path = SEQUENCES_DIR / sample["seq_id"] / "explanation.json"
    if sample_path.exists():
        with open(sample_path) as f:
            smp = json.load(f)
        print(f"\n📂 Sample explanation.json ({sample['seq_id']}):")
        print(json.dumps(smp, indent=2))
 
print("\n✅ Phase 6 complete — full pipeline end-to-end!")
print("\n🔥 Pipeline Summary:")
print("   Phase 1 → Dataset loading")
print("   Phase 2 → Sequence generation (valid / invalid)")
print("   Phase 3 → Feature extraction (angles, smoothness, shape)")
print("   Phase 4 → Rule-based validation (decision + violations)")
print("   Phase 5 → RAG retrieval (local embeddings + FAISS)")
print("   Phase 6 → LLM explanation (human-readable reasoning)  ✓")
```
